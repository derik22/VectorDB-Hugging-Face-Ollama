Question 1
What is a Vector Database (VectorDB) and how is it different from traditional databases?
Answer:
A Vector Database (VectorDB) is a specialized database designed to store, index, and retrieve vector embeddingsTraditional databases store structured data using rows and columns, while VectorDBs store numerical embeddings Key Differences:
• Traditional DB → exact keyword matching
• VectorDB → semantic similarity search
• Traditional DB → SQL queries
• VectorDB → nearest neighbor search
Examples: ChromaDB, Pinecone, Milvus, Weaviate, FAISS.  

Question 2

Answer:
1. ChromaDB → Lightweight and good for local AI applications.

3. Milvus → Enterprise-scale distributed VectorDB.
4. Weaviate → Semantic search and metadata filtering.
5. FAISS → Fast similarity search for research applications.
6. Qdrant → Real-time vector search for recommendation systems.

```python
# Question 6: ChromaDB Semantic Search

import chromadb
from sentence_transformers import SentenceTransformer

client = chromadb.Client()

collection = client.create_collection(name="documents")

model = SentenceTransformer('all-MiniLM-L6-v2')

docs = [
    "Artificial Intelligence is transforming industries",
    "Machine Learning enables predictive analysis",
    "Deep Learning uses neural networks"
]

embeddings = model.encode(docs).tolist()

collection.add(
    documents=docs,
    embeddings=embeddings,
    ids=["1", "2", "3"]
)

query = "AI applications"

query_embedding = model.encode([query]).tolist()

results = collection.query
(
    query_embeddings=query_embedding,
    n_results=2
)

print(results)
```

```python
# Question 7: Fine-tuning Hugging Face Model

from transformers import AutoTokenizer, AutoModelForCausalLM
from transformers import Trainer, TrainingArguments
from datasets import Dataset

model_name = "distilgpt2"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)

data = {
    "text": [
        "AI is changing the world",
        "Machine learning improves automation",
        "Generative AI creates content"
    ]
}

dataset = Dataset.from_dict(data)

def tokenize_function(example):
    return tokenizer(example["text"], truncation=True, padding="max_length")

tokenized_dataset = dataset.map(tokenize_function)

training_args = TrainingArguments(
    output_dir="./results",
    num_train_epochs=1,
    per_device_train_batch_size=2
)

trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_dataset
)

trainer.train()

print("Model fine-tuning completed.")
```

```python
# Question 9: Basic RAG System

import chromadb
from sentence_transformers import SentenceTransformer
import requests

client = chromadb.Client()
collection = client.create_collection("rag_docs")

model = SentenceTransformer('all-MiniLM-L6-v2')

docs = [
    "Python is a programming language",
    "Llama3 is a powerful open-source LLM",
    "RAG combines retrieval and generation"
]

embeddings = model.encode(docs).tolist()

collection.add(
    documents=docs,
    embeddings=embeddings,
    ids=["1", "2", "3"]
)

query = "What is RAG?"

query_embedding = model.encode([query]).tolist()

results = collection.query(
    query_embeddings=query_embedding,
    n_results=1
)

context = results['documents'][0][0]

prompt = f'''
Context: {context}

Question: {query}
Answer:
'''

response = requests.post(
    "http://localhost:11434/api/generate",
    json={
        "model": "llama3",
        "prompt": prompt,
        "stream": False
    }
)

print(response.json()["response"])
```

```python
# Question 10: Health-Tech Chatbot Architecture

import chromadb
from sentence_transformers import SentenceTransformer
import requests

client = chromadb.Client()
collection = client.create_collection("medical_docs")

model = SentenceTransformer('all-MiniLM-L6-v2')

docs = [
    "Diabetes is a chronic condition affecting blood sugar levels.",
    "Hypertension increases cardiovascular disease risk.",
    "Regular exercise improves heart health."
]

embeddings = model.encode(docs).tolist()

collection.add(
    documents=docs,
    embeddings=embeddings,
    ids=["1", "2", "3"]
)

query = "What causes high blood pressure?"

query_embedding = model.encode([query]).tolist()

results = collection.query(
    query_embeddings=query_embedding,
    n_results=2
)

context = "\n".join(results['documents'][0])

prompt = f'''
You are a medical AI assistant.

Context:
{context}

Question:
{query}

Answer:
'''

response = requests.post(
    "http://localhost:11434/api/generate",
    json={
        "model": "llama3",
        "prompt": prompt,
        "stream": False
    }
)

print(response.json()["response"])
```
